```python
base_dir = "Bidcoin"  # Bidcoin 폴더 경로 (상대/절대 경로로 수정 가능)

# Bidcoin 폴더 안의 json 파일 목록 가져오기
json_paths = glob.glob(os.path.join(base_dir, "*.json"))

dfs = dict() # category별 RAGAS 점수
for path in json_paths:
    df = pd.read_json(path)
    category = df['category'].iloc[0]
    ragas = pd.DataFrame([i for i in df['evaluation']]).fillna(0).mean().to_dict()
    dfs[category] = ragas
```

```python
df_a = pd.read_json(r"/home/spai0722/Bidcoin/eval_results_20260419_134319.json")
df_a
```

```python
for i in df_a['answer']:
    with open("result.txt", "a", encoding="utf-8") as f:
        f.write(i + df_a[df_a['answer'] == i]['category'].iloc[0] + "\n" 
                + " ".join([f"{k}:{v}" for k, v in df_a[df_a['answer'] == i]['evaluation'].iloc[0].items()]) + "\n")
```

```python
[f"{k}:{v}" for k, v in df_a[df_a['answer'] == i]['evaluation'].iloc[0].items()]
```

```python
import pandas as pd

# 1) evaluation에서 reason 제외하고 숫자 dict만 남기기
def drop_reason(d: dict) -> dict:
    if isinstance(d, dict):
        return {k: v for k, v in d.items() if k != "reason"}
    return d

df_a["eval_no_reason"] = df_a["evaluation"].apply(drop_reason)

# 2) dict를 여러 컬럼으로 펼치기
eval_df = pd.json_normalize(df_a["eval_no_reason"])
# 예: columns = ['relevance', 'faithfulness', 'correctness', 'completeness',
#                'context_precision', 'context_recall']
```

```python
eval_df
```

```python
# 3) 원본 df와 합치기
df_eval = pd.concat([df_a[["category"]], eval_df], axis=1)

# 4) category별 평균
category_mean = df_eval.groupby("category").mean(numeric_only=True)
df_b = pd.DataFrame(category_mean)
df_b
```

```python
df_b.to_csv("retrieval_experiments_summary2.csv", index=True, index_label="category")
```
