### hwp불러오기 정상화 확인

```python
from pathlib import Path
import subprocess
import pandas as pd

folder = Path("/home/shared/files")

target_files = sorted(folder.rglob("*.hwp"))

print(f"전체 HWP 파일 수: {len(target_files)}")
for f in target_files:
    print("-", f.name)

results = []

for f in target_files:
    exists_ok = f.exists()
    size_ok = f.stat().st_size if exists_ok else 0

    # 1) 바이너리 파일 자체는 열리는지
    binary_open_ok = False
    binary_error = ""
    try:
        with open(f, "rb") as fp:
            head = fp.read(64)
        binary_open_ok = True
    except Exception as e:
        binary_error = str(e)

    # 2) hwp5txt로 텍스트 추출까지 되는지
    extract_ok = False
    text_len = 0
    preview = ""
    extract_error = ""

    try:
        result = subprocess.run(
            ["hwp5txt", str(f)],
            capture_output=True,
            text=True,
            encoding="utf-8",
            errors="ignore"
        )

        text = result.stdout.strip()
        err = result.stderr.strip()

        if result.returncode == 0 and len(text) > 0:
            extract_ok = True
            text_len = len(text)
            preview = text[:300]
        else:
            extract_error = err if err else "(stdout 비었거나 returncode!=0)"
    except Exception as e:
        extract_error = str(e)

    results.append({
        "파일명": f.name,
        "파일존재": exists_ok,
        "파일크기(byte)": size_ok,
        "바이너리열기성공": binary_open_ok,
        "텍스트추출성공": extract_ok,
        "텍스트길이": text_len,
        "미리보기": preview,
        "바이너리에러": binary_error,
        "추출에러": extract_error[:500]
    })

df_check = pd.DataFrame(results)
df_check
```
